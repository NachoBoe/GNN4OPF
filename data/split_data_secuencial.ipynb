```python
import numpy as np

red = './reduru'
# Load input.npy file
input = np.load(red+'/input.npy')
# Load vm_pu_opt.npy
vm_pu_opt = np.load(red+'/vm_pu_opt.npy')

q_switch_shunt = np.load(red+'/q_switch_shunt_opt.npy')

print(input.shape)
print(vm_pu_opt.shape)
print(q_switch_shunt.shape)
```

```python
# Concatenate two arrays
# output = np.concatenate((input, vm_pu_opt), axis=2)
output = np.concatenate((input, vm_pu_opt, q_switch_shunt), axis=2)

# Do train, test and validation split
# np.random.shuffle(output)
train = output[0:int(output.shape[0]*0.77)] # Hasta abril de 2022
val = output[int(output.shape[0]*0.77):int(output.shape[0]*0.93)]
test = output[int(output.shape[0]*0.93):]
```

```python
# save train.npy
np.save(red+'_secuencial/train/input.npy', train[:, :, :4])
np.save(red+'_secuencial/train/vm_pu_opt.npy', train[:, :, 4:5])
np.save(red+'_secuencial/train/q_switch_shunt_opt.npy', train[:, :, 5:6])
np.save(red+'_secuencial/val/input.npy', val[:, :, :4])
np.save(red+'_secuencial/val/vm_pu_opt.npy', val[:, :, 4:5])
np.save(red+'_secuencial/val/q_switch_shunt_opt.npy', val[:, :, 5:6])
np.save(red+'_secuencial/test/input.npy', test[:, :, :4])
np.save(red+'_secuencial/test/vm_pu_opt.npy', test[:, :, 4:5])
np.save(red+'_secuencial/test/q_switch_shunt_opt.npy', test[:, :, 5:6])
```
